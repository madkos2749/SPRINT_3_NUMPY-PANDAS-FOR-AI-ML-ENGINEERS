# **DATE & TIME OPERATIONS**

Date and Time Operations are used to create, convert, extract, and manipulate dates and times in a dataset. In Python Pandas, to_datetime() converts values into date-time format.

```python
import pandas as pd
orders = pd.DataFrame({
    "Order_ID": ["01", "02", "03", "04", "05", "06"],
    "Customer": ["Anju", "Arha", "Arun", "Amala", "Ardhya", "Archana"],
    "Order_Date": [
        "2024-11-11",
        "2025-05-05",
        "2026-10-10",
        "2021-04-04",
        "2020-07-27",
        "2022-02-02"
    ],
    "Delivery_Date": [
        "2024-11-18",
        "2025-05-14",
        "2026-10-14",
        "2021-04-08",
        "2020-02-09",
        "2022-02-19"
    ],
    "Amount": [18000, 1200, 5656, 8945, 4563, 5897]
})

print(orders)

orders["Order_Date"] = pd.to_datetime(orders["Order_Date"])
orders["Delivery_Date"] = pd.to_datetime(orders["Delivery_Date"])
print("\n After converting to datetime:")
print(orders.dtypes)               
```

## DATE TIME

### DEFINITION:
Datetime represents both the date and time in a single value. In Python Pandas, pd.to_datetime() converts date and time values into a datetime format for easy manipulation and analysis.

```python
from datetime import datetime
current_datetime = datetime.now()

print(current_datetime)
```

```python
# Create a Specific Datetime

dt = datetime(1999, 4, 27, 12, 32, 33)

print(dt)
```

## TIMESTAMP

### DEFINITION:
A Timestamp in Pandas represents a specific date and time. It is used to store, compare, and perform operations on dates and times, such as extracting the year, month, day, or calculating time differences.

#### EXAMPLE:


```python
import pandas as pd
DOB_Stamp = pd.Timestamp("1996-11-3")
print("DOB TIMESTAMP needed is: ", DOB_Stamp)
```

```python
# TimeStamp with Time
import pandas as pd
Time_TimeStamp = pd.Timestamp("1999-10-3 3:30:33")
print("Time Stamp needed with Time is : ", Time_TimeStamp)
```

## DATEPARSING

### DEFINITION:
Date parsing is the process of converting date values stored as strings (text) into a proper date/time format that Pandas can understand and use for calculations, comparisons, and date operations.

For example, "11-11-2024" can be parsed into a date using pd.to_datetime().

#### EXAMPLE:


```python
import pandas as pd
date_value = pd.to_datetime("15-12-1999 5:50:55", dayfirst=True)

print(date_value)
```

```python
# Parsing a DataFrame Column

orders["Order_Date"] = pd.to_datetime(
    orders["Order_Date"]
)
print(orders["Order_Date"])
```

## TIME DELTA

### DEFINITION:
Timedelta represents the difference or duration between two dates or times. It is useful for calculating how many days an order takes to be delivered, how many hours have passed, or adding days to a date.

#### EXAMPLE:
* *For Example:*
  April 27 ---> April 30
  * The Difference is: 3 days.

```python
import pandas as pd
actual_deadline = pd.Timestamp("2026-10-20")
today_date = pd.Timestamp("2026-10-24")
difference = today_date - actual_deadline
print("Deadline Crossed and Exceeded Days are : ", difference)
```

```python
# Calculate Delivery Time

orders["Delivery_Days"] = (
    orders["Delivery_Date"] -
    orders["Order_Date"]
).dt.days

print(orders)
```

## TIMEZONES

### DEFINITION:
Timezones help represent and manage date and time values according to different geographical regions. In Pandas, we can assign a timezone to a datetime value and convert it to another timezone.

* For example, we can convert a time from UTC to Indian Standard Time (IST).
     * UTC
     * Asia/Kolkata
     * America/New_York
     * Europe/London
     * Asia/Tokyo

#### EXAMPLE:


```python
import pandas as pd
us_timestamp = pd.Timestamp(
    "2026-10-26 11:57:57",
    tz = "UTC"
)
india_timestamp = us_timestamp.tz_convert(
    "Asia/Kolkata"
)
print("INDIAN TIME ZONE is : ", india_timestamp)
```

## EXTRACTING DATE FEATURES

### DEFINITION:
Extracting date features means taking useful parts of a date, such as the year, month, day, or day of the week, from a datetime column. It helps analyze data by month, year, or day.

#### EXAMPLE:


```python
# Extract Year
orders["Year"] = orders["Order_Date"].dt.year
print("YEARS EXTRACTED are : ")
print(orders["Year"])
```

```python
# Extract Month
orders["Months"] = orders["Order_Date"].dt.month
print("MONTHS EXTRACTED are : ")
print(orders["Months"])
```

```python
# Extract DAYS
orders["Days"] = orders["Order_Date"].dt.day
print("DAYS EXTRACTED are : ")
print(orders["Days"])
```

```python
# Extract Day Name
orders["Day_Name"] = (
    orders["Order_Date"].dt.day_name()
)
print(" Extracted DAY NAMES are : ")
print(orders["Day_Name"])
```

```python
# Extract Day of Week
orders["Day_of_Week"] = (
    orders["Order_Date"].dt.dayofweek
)
print("Extracted DAYS OF WEEK is : ")
print(orders["Day_of_Week"])
```

* **Extract Multiple Features**
   * Instead of creating each feature seperately, we can create all the required features in a single cell.

```python
orders["Year"] = orders["Order_Date"].dt.year
orders["Month"] = orders["Order_Date"].dt.month
orders["Day"] = orders["Order_Date"].dt.day
orders["Quarter"] = orders["Order_Date"].dt.quarter
orders["Day_of_Week"] = orders["Order_Date"].dt.dayofweek
orders["Is_Weekend"] = orders["Order_Date"].dt.dayofweek >= 5

print(orders)
```
