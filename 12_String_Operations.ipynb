# **STRING OPERATIONS**

* String operations are used to manipulate and analyze text data in a Pandas DataFrame. They help us convert text to uppercase or lowercase, find text, replace words, and extract parts of strings.

* In Pandas, we use the .str accessor to perform string operations on a column.

### EXAMPLE:


```python
import pandas as pd
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "NAME": [
        " Amrutha",
        "Gokul",
        "PRATUSHA",
        "Ratna",
        "Lohith"
    ],
    "EMAIL": [
        "amrutha@gmail.com",
        "gokul@yahoo.com",
        "pratusha@gmail.com",
        "ratna@outlook.com",
        "lohith@gmail.com"
    ],
    "CITY": [
        "Pune",
        "Bopal",
        "HUBLI",
        "MUMBAI",
        "OOTY"
    ],
    "PHONE": [
        "7890123456",
        "8901234567",
        "9012345678",
        "9988776655",
        "9080706050"
    ]
})

print(customers)
```

## **lower()**

### DEFINITION:
* The lower() method converts all uppercase letters in a string into lowercase letters. In Pandas, we use .str.lower() to convert all text values in a column to lowercase.

* It is useful for standardizing customer names, email addresses, and city names.

```python
customers["NAME"].str.lower()
```

## **upper()**

### DEFINITION:
* The upper() method converts all lowercase letters in a string into uppercase letters. In Pandas, we use .str.upper() to convert all text values in a column to uppercase.

* It is useful for standardizing customer names, email addresses, and city names.

```python
customers["NAME"].str.upper()
```

## **strip()**

### DEFINITION:
* The strip() method removes unwanted spaces or specified characters from the beginning and end of a string. In Pandas, we use .str.strip() to remove extra spaces from every string in a column.

* It is useful for cleaning customer names, city names, and other text data.

```python
customers["CITY"].str.strip()
```

## **strip()**

### DEFINITION:
The strip() method removes extra spaces or specified characters from the beginning and end of a string. In Pandas, we use .str.strip() to clean text values in a DataFrame column.

```python
customers["EMAIL"].str.strip()
```

## **split()**

### DEFINITION:
* The split() method divides a string into smaller parts based on a separator, such as a space, comma, or @ symbol.
* In Pandas, we use .str.split() to split text values in a column.

* It is useful for separating full names, extracting email providers, and dividing text into individual words.

```python
customers["CITY"].str.split()
```

## **contains()**

### DEFINITION:
* The contains() method checks whether a string contains a specific word, letter, or pattern. In Pandas,
* we use .str.contains() to search for text in a column. It returns True if the pattern is found and False otherwise.

* It is useful for filtering customer names, searching email addresses, and finding specific text in a DataFrame.

```python
customers["EMAIL"].str.contains("yahoo")
```

## **replace()**

### DEFINITION:
* The replace() method is used to find specific text in a string and replace it with another value. In Pandas,
* we use .str.replace() to replace text in every value of a column.

* It is useful for correcting spelling mistakes, standardizing city names, and cleaning customer information.

```python
cities = pd.Series([
    "Chennai",
    "Madurai",
    "Chennai",
    "Madras"
])
cities = cities.str.replace(
    "Madras",
    "Chennai"
)

print(cities)
```

## **Regular Expressions [Regrex]**

### DEFINITION:

Regular Expressions (Regex) are patterns used to search, match, extract, or replace specific text in a string. In Pandas, we can use Regex with string operations such as .str.contains(), .str.extract(), and .str.replace().

Regex is useful for finding email addresses, checking phone numbers, and removing unwanted characters from text data.

```python
names = pd.Series([
    "Amrutha@8910",
    "Ruthwik&765",
    "Usha$789"
])

clean_names = names.str.replace(
    r"[^A-Za-z0-9]",
    "",
    regex=True
)

print(clean_names)
```

* **Regex -: Find words containing Pattern.**

```python
products = pd.Series([
    "iPhone",
    "Samsung Phone",
    "Laptop",
    "Phone Case",
    "Realme",
    "Redmi"
])

products.str.contains(
    r"phone",
    case=False,
    regex=True
)
```
