## **1. Create a Numpy array**

**ANS:**
### DEFINITION:
A NumPy array is a data structure used to store multiple values of the same data type in a single variable. NumPy arrays support fast mathematical calculations and efficient data processing.

```python
import numpy as np
#Create a Numpy array of Different Temperatures.
temperatures = np.array([17, 19, 21, 23, 25])
print("Daily Temperatures :", temperatures)
print("First Temperatures :", temperatures[0])
print("Average Temperature:", np.mean(temperatures))
```

## **2. Explain the difference between a NumPy Array and a Python List ?**

**ANS:**

### **Differences:**

* **Python List**

  * Built into Python.
  * Can store different data types.
  * Mathematical operations often require loops or comprehensions.
  * Generally less efficient for large numerical calculations.
  * Created using [].

* **Numpy Array**

   * Requires the NumPy library.
   * Usually stores elements of the same data type.
   * Supports vectorized mathematical operations.
   * Generally faster for large numerical calculations.
   * Created using np.array().

```python
import numpy as np

# Python List
list_values = [10, 20, 30]
print("Python List:", list_values)
print("List + 5:", [x + 5 for x in list_values])

# NumPy Array
array_values = np.array([10, 20, 30])
print("NumPy Array:", array_values)
print("Array + 5:", array_values + 5)
```

## **4. Difference between Series and DataFrame ?**

**ANS:**

**SERIES:**
  * A Series is one-dimensional.
  * It contains a single column of data.
  * It uses a single index.
  * Example: A list of marks.

**DataFrame :**
  * A DataFrame is two-dimensional.
  * It contains multiple columns of data.
  * It uses row and column labels.
  *Example: Student names and marks.

```python
import pandas as pd

# Series example
marks = pd.Series([80, 90, 75])
print("Series:")
print(marks)

# DataFrame example
students = pd.DataFrame({
    "Name": ["Rajesh", "Sujatha", "Rajini"],
    "Marks": [80, 90, 75]
})
print("DataFrame:")
print(students)
```

## **6. Merge two DataFrames ?**

**ANS:**
Merging two DataFrames means combining data from two DataFrames using a common column. In Pandas, the merge() function is used for this purpose.

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2],
    "Name": ["Ranjith", "Sitara"]
})

df2 = pd.DataFrame({
    "ID": [1, 2],
    "Marks": [85, 90]
})

result = pd.merge(df1, df2, on="ID")
print(result)
```

## **7. Perform GroupBy and Aggregation ?**

**ANS:**

*groupby()* is used to group data based on a column, and aggregation is used to calculate values such as sum, average, or count for each group.

```python
import pandas as pd

data = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [20000, 15000, 30000, 25000]
})

result = data.groupby("Department")["Salary"].sum()
print(result)
```

## **8. Filter Rows Based on Multiple Conditions ?**

**ANS:**

 In Pandas, filtering rows based on multiple conditions means selecting rows that satisfy two or more conditions using operators like & (AND) and | (OR).

```python
import pandas as pd

data = pd.DataFrame({
    "Name": ["Asha", "Balu", "Chinnu"],
    "Marks": [80, 50, 90],
    "Age": [18, 20, 19]
})

result = data[(data["Marks"] > 60) & (data["Age"] < 20)]
print(result)
```

## 10. Calculate statistical measures using NumPy ?**

**ANS:**

NumPy is a Python library used to perform mathematical and statistical calculations such as mean, median, standard deviation, and maximum.

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", round(np.std(marks), 2))
```

## **11. Optimize a DataFrame for memory usage ?**

**ANS:**

Optimizing a DataFrame means reducing the amount of memory used to store data. In Pandas, we can use smaller data types like int8 instead of int64 to save memory.

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [18, 20, 22],
    "Marks": [70, 80, 90]
})

print("Before optimization:")
print(data.memory_usage(deep=True))

data["Age"] = data["Age"].astype("int8")
data["Marks"] = data["Marks"].astype("int8")

print("After optimization:")
print(data.memory_usage(deep=True))
```
