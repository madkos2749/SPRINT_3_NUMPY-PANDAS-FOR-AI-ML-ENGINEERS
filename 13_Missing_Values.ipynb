# **MISSING VALUES**

* Missing values are data entries that are empty or unavailable in a DataFrame. In Pandas, missing values are commonly represented by NaN (Not a Number) or None.

* Missing values can occur when information is not provided, data is lost, or records are incomplete. Pandas provides methods to identify, remove, and fill these values.

```python
import numpy as np

employees = pd.DataFrame({
    "EMPLOYEE_ID" : [np.nan, "STK_263_15222", "STK_143_7857", np.nan, "STK_897_2362"],
    "NAME" : ["Hema", "Anshu", "Eswar", "Rohan", "Hanish"],
    "DOJ" : [np.nan, "2008-8-8", "2009-9-9", "2007-7-7", "2006-6-6"],
    "ATTENDANCE" : [88.8, 99.9, 77.7, np.nan, np.nan]
})
print(employees)
```

## isnull()

### DEFINITION:

* The isnull() method is used to identify missing values in a Pandas Series or DataFrame.
* It returns True if a value is missing and False if a value is present.

* It is useful for finding empty values and checking data quality before analysis.
    * True --> Missing
    * False --> Not Missing

```python
print(employees.isnull())
```

```python
# Count Missing Values

print(employees.isnull().sum())
```

```python
# Total Missing Values

total_missing = employees.isnull().sum().sum()

print(total_missing)
```

## notnull()

### DEFINITION:
* The notnull() method is used to identify values that are present and not missing in a Pandas Series or DataFrame.
* It returns True if a value exists and False if the value is missing.

* It is useful for finding complete records and filtering rows that contain valid data.

```python
print(employees.notnull())
```

# dropna()

### DEFINITION:
* The dropna() method is used to remove rows or columns that contain missing values (NaN or None) from a Pandas DataFrame.

* By default, dropna() removes rows containing at least one missing value. It is useful for cleaning incomplete data before analysis.

```python
clear_employees = (employees.dropna())
print(clear_employees)
```

* **dropna(axis = 1)**
   * In **dropna(axis = 1)**; We can remove columns containing missing values.

```python
final_employees = (employees.dropna(axis = 1))
print("FINAL EMPLOYEES are: ")
print(final_employees)
```

## fillna()

### DEFINITION:
* The fillna() method is used to replace missing values (NaN or None) in a Pandas Series or DataFrame with a specified value, such as 0, a text value, or the mean of a column.

* It is useful when you want to keep records instead of deleting rows containing missing data.

```python
employees["EMPLOYEE_ID"].fillna(1)
```

```python
# Fill with Mean

MEAN_ATTENDANCE = employees["ATTENDANCE"].mean()

employees["ATTENDANCE"] = employees["ATTENDANCE"].fillna(MEAN_ATTENDANCE)
print(employees)
```

## interpolate()

### DEFINITION:
* The interpolate() method is used to estimate and fill missing numerical values based on existing values.
* By default, Pandas uses linear interpolation, which calculates missing values by assuming a straight-line change between known values.

* It is useful for filling gaps in temperature readings, sales data, sensor measurements, and time-series data.

```python
SALES = pd.DataFrame({
    "DAY": [
        "MONDAY",
        "TUESDAY",
        "WEDNESDAY",
        "THURSDAY",
        "FRIDAY",
        "SATURDAY",
        "SUNDAY"
    ],
    "AMOUNT": [
        88888,
        55000,
        66000,
        np.nan,
        26999,
        99999,
        77777
    ]
})

print(SALES)
```

* **LINEAR INTERPOLATION**

```python
SALES["AMOUNT"] = (
    SALES["AMOUNT"]
    .interpolate()
)

print(SALES)
```

## MISSING VALUE HANDLING STRATEGIES

### DEFINITION:

* Missing value handling strategies are methods used to identify, remove, or replace missing data in a DataFrame. They help improve data quality and make analysis more reliable.

* The best strategy depends on the amount of missing data and the type of information stored in the column.

* Common missing value handling strategies
  * **1. Detect missing values** — Use isnull() to identify missing entries.

  * **2. Remove missing values** — Use dropna() to delete incomplete rows or columns.

  * **3. Fill with a fixed value** — Use fillna() to replace missing values with 0, "Unknown", or another suitable value.

  * **4. Fill with mean or median** — Replace missing numerical values with the column's mean or median.

  * **5. Fill with mode** — Replace missing categorical values with the most frequent value.

  * **6. Interpolation** — Use interpolate() to estimate missing numerical values from surrounding values.
