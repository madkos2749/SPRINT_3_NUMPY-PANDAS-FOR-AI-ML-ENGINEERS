# NUMPY RANDOM MODULE

### DEFINITION:
The NumPy Random Module is a module in NumPy used to generate random numbers and random data. It provides functions such as np.random.randint(), np.random.rand(), and np.random.choice() for generating random integers, random floating-point numbers, and random selections.

## RANDOM NUMBERS

### DEFINITION:

Random numbers are numbers generated unpredictably within a specified range. In NumPy, the Random Module provides functions such as np.random.randint() to generate random integers.

* **SYNTAX :**
  * np.random.randint()

#### EXAMPLE:
Generating a Random Lottery Number.

```python
import numpy as np
number = np.random.randint(1, 111)
print(" LOTTERY NUMBER is : ", number)
```

## RANDOM SEED

### DEFINITION:

A random seed is a starting value used to generate a sequence of pseudo-random numbers. In NumPy, setting a seed allows us to generate the same sequence of random numbers every time the program runs.

#### EXAMPLE:

Simulating Daily Sales.

```python
import numpy as np
np.random.seed(10)
sales = np.random.randint(11, 555, 55)
print("DAILY SALES : ", sales)
```

## RANDOM INTEGERS

### DEFINITION:

Random integers are whole numbers generated randomly within a specified range. In NumPy, the np.random.randint() function is used to generate random integers.

* **SYNTAX :**
     np.random.randint(low, high)

  Here, low is the starting value (inclusive) and high is the upper limit (exclusive).

#### EXAMPLE:

Generating Random Queue Tokens.

```python
import numpy as np
tokens = np.random.randint(3, 100, 5)
print(" RANDOM QUEUE TOKENS are : ", tokens)
```

## RANDOM CHOICE

### DEFINITION:

The np.random.choice() function in NumPy is used to select a random element from a given array or sequence.

* **SYNTAX :**
      np.random.choice(array)

#### EXAMPLE:
Randomly Assigning a Dessert Menu.

```python
import numpy as np
desserts = ["Apricot Delight", "Tiramasu", "Kaju Banana", "Jack Fudge", "Custard Smooch"]
print("RANDOM DESSERTS are : ", np.random.choice(desserts,2))
```

## SHUFFLE

### DEFINITION:

The np.random.shuffle() function in NumPy is used to randomly rearrange the elements of an array. It modifies the original array directly.

* **SYNTAX :**

  np.random.shuffle(array)

#### EXAMPLE:
Shuffling Music List.

```python
import numpy as np
music_list = np.array(["Melody Songs", "Devotional Songs", "Item Songs", "Folk Songs", "Rap Songs"])
np.random.shuffle(music_list)
print(" NEW MUSIC LIST is : ", music_list)
```

## PERMUTATION

### DEFINITION:

permutation() shuffles elements and returns a new arrangement without modifying the original array.
The output order may change each time the program runs, and the original array remains unchanged.

* **SYNTAX :**
           numpy.random.permutation()

#### EXAMPLE:


```python
import numpy as np
P = np.array([26, 59, 35, 42, 76])
print(" NEW PERMUTATION LIST is : ", np.random.permutation(P))
```

## RANDOM DISTRIBUTIONS

### DEFINITION:

Random distributions in NumPy are used to generate random numbers based on a specific probability distribution. NumPy provides functions such as normal(), binomial(), and uniform().

#### EXAMPLE:

```python
import numpy as np
rd = np.random.normal(89, 64, 2)
bd = np.random.binomial(16, 0.7, 3)
ud = np.random.uniform(13, 14, 1)
print("RD = ", rd)
print("BD = ", bd)
print("UD = ", ud)
```
