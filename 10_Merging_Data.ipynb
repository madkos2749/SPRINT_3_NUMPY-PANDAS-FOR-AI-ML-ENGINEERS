# **COMBINING DATASETS**

Combining datasets is the process of joining two or more datasets into a single dataset to make data analysis easier. In Python Pandas, concat() and merge() are commonly used for combining datasets.

#### EXAMPLE:
Create Sample Dataset1

```python
import pandas as pd
students = pd.DataFrame({
    'Name': ['Ravi', 'Sita'],
    'Marks': [80, 90],
    'Branch': ['DS', 'ML']
})

print(students)
```

#### EXAMPLE:
Create Sample Dataset2

```python
import pandas as pd
Department = pd.DataFrame({
    'Branch': ['AI', 'DS'],
    'Branch_id': [505,707]
})
print(Department)
```

#### EXAMPLE:
Create Sample Dataset3

```python
import pandas as pd
new_students = pd.DataFrame({
    'Name': ['Kamala', 'Maharsh'],
    'Branch': ['ML', 'AI'],
    'Marks': [88,99]
})
print(new_students)
```

## CONCAT()

### DEFINITION:

concat() is a Pandas function used to combine two or more datasets by placing them vertically (rows) or horizontally (columns).

#### EXAMPLE:


```python
combined_students = pd.concat(
    [students, new_students],
    ignore_index=True
)

print(combined_students)
```

```python
children_info = pd.DataFrame({
    "Name": ["Gagan", "Ganga", "Prudvi"],
    "Age": [22, 23, 21]
})

student_scores = pd.DataFrame({
    "Marks": [95, 88, 91],
    "Grade": ["A", "B", "A"]
})

result = pd.concat(
    [children_info, student_scores],
    axis=1
)

print(result)
```

## MERGE()

### DEFINITION:
merge() is a Pandas function used to combine two datasets based on a common column or key.

#### EXAMPLE:

```python
merged_data = pd.merge(
    students,
    Department,
    on="Branch"
)

print(merged_data)
```

Types of Merge
Pandas supports:

* Inner Join
* Left Join
* Right Join
* Outer Join

* **inner join**
* Returns Matching Records.

```python
pd.merge(
    students,
    Department,
    on="Branch",
    how="inner"
)
```

* **left join()**

```python
pd.merge(
    students,
    Department,
    on="Branch",
    how="left"
)
```

* **right join()**

```python
pd.merge(
    students,
    Department,
    on="Branch",
    how="right"
)
```

* **outer join()**

```python
pd.merge(
    students,
    new_students,
    on="Marks",
    how="outer"
)
```

## JOIN()

### DEFINITION:

join() is a Pandas method used to combine two datasets based on their index by default. It is useful for bringing related columns together.

```python
student_info = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Prudvi"],
    "Department": ["AI", "ML", "DS"]
}, index=[101, 102, 103])

student_marks = pd.DataFrame({
    "Marks": [95, 88, 91]
}, index=[101, 102, 103])
```

```python
result = student_info.join(student_marks)

print(result)
```

## **append()**

### DEFINITION:
append() is used to add new rows or records to an existing dataset.

Note: In older versions of Pandas, DataFrame.append() was used, but it was removed in Pandas 2.0.

* append() was a Pandas method used to add new rows to an existing DataFrame
* Deprecation of append()
DataFrame.append() was deprecated in Pandas 1.4.0.

It was removed in Pandas 2.0, so the old code will not work in Pandas 2.0 or later.

The recommended alternative is pd.concat().

#### EXAMPLE:


```python
result = pd.concat(
    [students, new_students],
    ignore_index=True
)

print(result)
```

## **combine_first()**

### DEFINITION:
combine_first() is a Pandas method used to combine two datasets by filling missing (NaN) values in the first dataset with corresponding values from the second dataset.

```python
#### EXAMPLE:
student_data1 = pd.DataFrame({
    "Name": ["Amulya", "Rama", "Pardhu"],
    "Marks": [59, None, 87]
})

student_data2 = pd.DataFrame({
    "Name": ["Amruth", "Kushi", "Omkar"],
    "Marks": [None, 88, 92]
})
```

```python
result = student_data1.combine_first(student_data2)

print(result)
```
