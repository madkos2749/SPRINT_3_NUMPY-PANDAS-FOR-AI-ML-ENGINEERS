# **INTRODUCTION TO PANDAS**

## Whai is PANDAS ?

Pandas is a popular Python library used for data analysis, data manipulation, and data cleaning. It helps you organize, process, analyze, and understand data easily.

* For example, if you have student marks, employee salaries, sales records, or Excel files, you can use Pandas to read the data, filter it, calculate totals, and find useful information.

Pandas makes working with structured data easier:
* Analyze large amounts of data.
*  Handle missing values and duplicate records.
*  Adds, remove, filter and modifies data.
*  Read Excel, CSV, and other supported file formats.

#### EXAMPLE:
Create a table containing student names and marks

```python
import pandas as pd
data = {
    "NAMES": ["HaSa", "Mad", "Hema", "Anj", "Josh", "Chand"], 
    "SURNAMES": ["Alu", "Kos", "Anam", "Anabh", "Jonna", "Dhula"]
}
df = pd.DataFrame(data)
print(df)
```

## Why PANDAS ?

Pandas is used in Python because it makes working with data easy, fast, and efficient. It provides simple functions to read, organize, clean, modify, and analyze data without writing complicated code.

* **1. Easy data handling**

Pandas organizes data into rows and columns, similar to an Excel spreadsheet.

* **2. Saves time**

You can process thousands or millions of rows efficiently without manually checking each record.

* **3. Easy filtering and sorting**

Find specific records, arrange data, and select rows based on conditions.

* **4. Data cleaning**

Handle missing values, remove duplicates, and correct data formats.

* **5. Easy calculations**

Calculate totals, averages, minimums, maximums, and other statistics with simple functions.

* **6. Supports multiple file formats**

Read and write CSV, Excel, JSON, and SQL data, among other supported formats.

### EXAMPLE:

want to find students who scored more than 80 and calculate the average marks.

```python
import pandas as pd
STUDENTS = {
    "Names" : ["Joshika", "Chandana", "Harshita", "Samyu", "Prabha"], 
    "Marks" : [82, 91, 64, 99, 27]
}
df= pd.DataFrame(STUDENTS)
# Filter students who scored above 80
print(df[df["Marks"] > 80])

# Calculate average marks
print("AVERAGE is : ", df["Marks"].mean())
```

## SERIES

### DEFINITION:

A Series in Pandas is a one-dimensional labeled data structure that stores a sequence of values, where each value has an index.

It can store:
* Numbers
* Strings
* Boolean values
Each value has an index.

### EXAMPLE:


```python
import pandas as pd
FRUITS = pd.Series(['Passion', 'Apricot', 'Lime', 'Litchie', 'Custard'])
print(FRUITS)
```

## DATA FRAME

### DEFINITION:

A DataFrame in Pandas is a two-dimensional labeled data structure consisting of rows and columns. It can store different types of data, such as integers, strings, and floating-point numbers.


#### EXAMPLE:

```python
import pandas as pd
WEATHER = {
    'CITY' : ['Mangalagir', 'Guntur', 'Vijayawada', 'Hyderabad', 'Coimbatore', 'Chennai', 'Salem'],
    'TEMPERATURE' : [45.4, 44.4, 45.4, 39.9, 38.3, 40.4, 37.7]
}
df = pd.DataFrame(WEATHER)
print("WEATHER REPORT is : ")
print(df)
```

## CREATING DATA FRAMES

### DEFINITION:

Creating a DataFrame in Pandas means converting data into a two-dimensional structure containing rows and columns using the pd.DataFrame() function.

* **SYNTAX :**
  pd.DataFrame()

#### EXAMPLE:

```python
import pandas as pd
Product_Data = {
    'ITEM' : ['Umbrella', 'Shower', 'Claw Clip', 'Dish Washer'],
    'PRICE' : [321, 555, 111, 9999]
}
PRODUCT = pd.DataFrame(Product_Data)
print("PRODUCTS DATA is : ")
print(PRODUCT)
```

## DATA TYPES

### DEFINITION:

Data types in Pandas specify the type of values stored in a Series or DataFrame column, such as integers, floating-point numbers, strings, and Boolean values.

#### EXAMPLE:

```python
import pandas as pd

data = {
    'Age': [20, 25, 30],
    'Salary': [15000.5, 20000.0, 25000.5],
    'Passed': [True, False, True]
}

df = pd.DataFrame(data)
print("DATA TYPES of the Given DATA are: ")
print(df.dtypes)
```

## INDEX

### DEFINITION:

An Index in Pandas is a label used to identify and access rows or elements in a Series or DataFrame. By default, Pandas assigns integer index labels starting from 0.

#### EXAMPLE:


```python
import pandas as pd
s = pd.Series([45, 60, 75], 
              index=['A', 'B', 'C'])

print(s)
print("Value at B:", s['B'])
```

## COLUMNS

### DEFINITION:

Columns in Pandas are vertical sections of a DataFrame that represent different variables or attributes. Each column has a name and can store a particular type of data.

#### EXAMPLE:

```python
import pandas as pd
data = {
    'Product': ['Pen', 'Pencil', 'Eraser'],
    'Quantity': [10, 20, 15]
}

df = pd.DataFrame(data)

print(df.columns)
```

## READING DATA

### DEFINITION:

#### EXAMPLE:
Suppose a CSV file named students.csv contains student details.

```python
import pandas as pd
data = {
    'Name' : ['Madhuri', 'Kosanam'],
    'Age' : [12, 21],
    'Course' : ['Python', 'SQL']
}

df = pd.DataFrame(data)
print(df)
df.to_csv('students.csv', index= False)
print("CSV file created")
```
