# **GROUPING & AGGREGATION**

* 1. **GROUPING**
Grouping is the process of organizing data into groups based on common values in one or more columns.

* 2. **AGGREGATION**
Aggregation is the process of summarizing data using calculations such as sum, average, count, minimum, and maximum.

```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101,102,103,104,105,106,107,108],
    "Name": ["Bhaskar","Balu","Baruni","Bhuvesh","Barath","Brindha","Beema","Bushan"],
    "Department": ["AI","ML","AI","DS","ML","DS","AI","ML"],
    "Gender": ["Male","Male","Female","Male","Male","Female","Female","Male"],
    "Marks": [76,88,99,94,82,72,89,88],
    "Age": [22,23,21,24,22,23,21,21]
})

print(employees)
```

## GROUPBY()

### DEFINITION:
groupby() is a Pandas function used to group data based on common values in a column, so that calculations can be performed on each group separately.

```python
employees.groupby("Department")["Marks"].mean()
```

## AGG()

### DEFINITION:
agg() is a Pandas function used to perform one or more aggregation operations, such as sum, mean, count, minimum, and maximum, on data.

#### EXAMPLE:


```python
employees.groupby("Department")["Marks"].agg(
    ["mean","max","min","count"]
)
```

## transform()

### DEFINITION:
transform() is a Pandas function that performs a calculation on each group and returns results with the same number of rows as the original data.

#### EXAMPLE:


```python
employees["Department_Average"] = employees.groupby(
    "Department"
)["Marks"].transform("mean")

print(employees)
```

## filter()

### DEFINITION:
filter() is a Pandas function used with groupby() to select or remove entire groups based on a specified condition.

#### EXAMPLE:

```python
employees.groupby("Department").filter(
    lambda x: x["Marks"].mean() > 90
)
```

## Pivot TABLES

### DEFINITION:

Pivot tables are used to summarize and organize data by grouping rows and columns and applying calculations such as sum, mean, or count.

#### EXAMPLE:

```python
pd.pivot_table(
    employees,
    values="Marks",
    index="Department",
    aggfunc="mean"
)
```

```python
# Multiple Aggregations
pd.pivot_table(
    employees,
    values="Marks",
    index="Department",
    aggfunc=["mean","max","min"]
)
```

## CROSSTAB

### DEFINITION:

crosstab() is a Pandas function used to calculate the frequency (count) of combinations between two or more columns.

#### EXAMPLE:

```python
pd.crosstab(
    employees["Department"],
    employees["Gender"]
)
```

```python
# Crosstab with Totals
pd.crosstab(
    employees["Department"],
    employees["Gender"],
    margins=True
)
```

```python
# Crosstab with Percentages
pd.crosstab(
    employees["Department"],
    employees["Gender"],
    normalize=True
)
```
