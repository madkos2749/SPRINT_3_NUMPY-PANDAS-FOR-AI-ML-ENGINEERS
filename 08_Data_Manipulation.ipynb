# **DATA MANIPULATION**

Data manipulation is the process of modifying, organizing, arranging, and transforming data to make it useful for analysis and decision-making.

#### EXAMPLE:
Create a Sample Dataset.

```python
import pandas as pd

fruits = pd.DataFrame({
    "Fruit_ID": [101, 102, 103, 104, 105, 105],
    "NAME": ["Grapes", "Muskmelon", "Watermelon", "Strawberry", "Raspberry", "Cranberry"],
    "QUALITY": ["Fresh", "Ripe", "Rotten", "Fresh", "Fresh", "Ripe"],
    "TASTE": ["Sour", "Sweet", "Sweet", "Sour", "Sweet", "sweet"],
    "QUANTITY": [11, 2, 3, 50, 41, 77]
})

print("FRUITS DATA")
print(fruits)
```

## RENAME COLUMNS

### DEFINITION:
RENAME COLUMNS means changing the name of an existing column in a table without changing the data stored in it.

In SQL, the syntax depends on the database system.

#### EXAMPLE:
Renaming Column name Percentage to Quantity.

```python
fruits.rename(columns={"QUANTITY":"PERCENTAGE"})
```

## DROP ROWS

### DEFINITION:

Drop Rows means removing unwanted rows (records) from a dataset or table. In SQL, we use the DELETE command to remove specific rows based on a condition.

#### EXAMPLE:

Deleting row

```python
fruits.drop(1)
```

* Deleting Multiple Rows.

```python
fruits.drop([1,3])
```

## DROP COLUMNS

### DEFINITION:
Drop Columns means removing an unwanted column from a table or dataset. In SQL, the ALTER TABLE command with DROP COLUMN is used to remove a column and all its stored values.

#### EXAMPLE:
Deleting column from dataset.

## DROP COLUMNS

### DEFINITION:
Drop Columns means removing an unwanted column from a table or dataset. In SQL, the ALTER TABLE command with DROP COLUMN is used to remove a column and all its stored values.

#### EXAMPLE:

Deleting Column from Dataset.

```python
fruits.drop(columns=["QUANTITY"])
```

## INSERT COLUMNS

### DEFINITION:
Insert Columns means adding a new column to an existing table or dataset. In SQL, the ALTER TABLE command with ADD COLUMN is used to create a new column.

#### EXAMPLE:
Adding new Column to Dataset.

```python
fruits.insert(5, "STATUs", ['Shipped', 'Packed', 'Delivered', 'Cancelled', 'Returned', 'Preparing'])
print(fruits)
```

## ASSIGN()

### DEFINITION:
ASSIGN means giving a value to a variable or creating a new column by assigning values or calculations to it. In data manipulation using Python Pandas, assignment is commonly used to add a new column or update an existing column.

#### EXAMPLE:


```python
FRUITS_new = fruits.assign(
    Pass=fruits["QUALITY"] == 'Fresh'
)
print("UPDATED FRUITS LIST is: ")
print(FRUITS_new)
```

## APPLY()

### DEFINITION:
apply() is a function in Python Pandas used to apply a function to each element in a Series or along rows or columns of a DataFrame.

#### EXAMPLE:


```python
fruits["NAME"].apply(str.upper)
```

## MAP()

### DEFINITION:
map() is a function in Python Pandas used to replace or transform each value in a Series (a single column) using a function or a dictionary.

#### EXAMPLE:


```python
grade_map = {
    "Shipped": "Excellent",
    "Packed": "Good",
    "Delivered": "Average",
    "Cancelled": "Worst",
    "Returned": "Fine",
    "Preparing": "Going ON"
}

fruits["RATING"] = fruits["STATUs"].map(grade_map)

print(fruits)
```

## REPLACE()

### DEFINITION:
replace() is a function in Python Pandas used to replace existing values in a DataFrame or Series with new values.

#### EXAMPLE:

```python
fruits = fruits.replace({
    "NAME":{
        "Grapes": "KiWi",
        "Muskmelon": "Dragon"
    }
})

print(fruits)
```

## DUPLICATE HANDLING

### DEFINITION:
Duplicate handling is the process of identifying and removing repeated records from a dataset to maintain data accuracy and avoid unnecessary repetition.

#### EXAMPLE:
* **DUPLICATE**

```python
fruits.duplicated()
```

* **COUNT Duplicate Rows**

```python
fruits.duplicated().sum()
```

* **REMOVE Duplicate Rows**

```python
fruits.drop_duplicates()
```
