# DATA INSPECTION

* Data Inspection is the process of examining a dataset to understand its structure, size, data types, missing values, and contents before performing data analysis or processing.

* **Purpose**: To understand data quality and identify errors or missing values before analysis.

### EXAMPLE:


```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df.head())       # View first 5 rows
print(df.info())       # Check columns and data types
print(df.isnull().sum())  # Check missing values
```

## HEAD()

### DEFINITION:

The head() function in Python's Pandas library is used to display the first few rows of a dataset. By default, it displays the first 5 rows. It helps us quickly inspect the data.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Fruit": [ "Kiwi", "Jack", "Muskmelon", "Watermelon", "Raspberry", "Strawberry", "Apple", "Mango", "Banana", "Orange", "Grapes"],
    "Price": [100, 80, 40, 60, 120, 90, 180, 800, 200, 1200, 660,]
}

df = pd.DataFrame(data)

print(df.head(4))
```

## TAIL()

### DEFINITION:
* The tail() function in Python's Pandas library is used to display the last few rows of a dataset. By default, it displays the last 5 rows.
* It helps us quickly inspect the ending records of a dataset.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "City": ["Delhi", "Mumbai", "Chennai", "Kolkata", "Pune", "Hyderabad", "Vijayawada", "Guntur"],
    "Temperature": [32, 30, 35, 28, 33, 29, 33, 31]
}

df = pd.DataFrame(data)

print(df.tail(3))
```

## SAMPLE()

### DEFINNITION:
The sample() function in Python's Pandas library is used to select random rows from a dataset. It helps inspect different records without viewing the entire dataset.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Animal": ["Lion", "Tiger", "Elephant", "Zebra", "Giraffe"],
    "Weight": [190, 220, 500, 300, 800]
}

df = pd.DataFrame(data)

print(df.sample(2, random_state=1))
```

## INFO()

### DEFINITION:

The info() method in Python's Pandas library provides a summary of a DataFrame. It displays the number of rows and columns, column names, data types, non-null values, and memory usage.

### EXAMPLE:

```python
import pandas as pd

data = {
    "Book": ["Python", "Java", "SQL"],
    "Price": [250, 300, 200],
    "Available": [True, False, True]
}

df = pd.DataFrame(data)

df.info()
```

## DESCRIBE()

### DEFINITION:
The describe() method in Python's Pandas library is used to generate summary statistics for a dataset. By default, it provides statistics for numerical columns, including count, mean, standard deviation, minimum, maximum, and quartile values.

### EXAMPLE:


```python
import pandas as pd

data = {
    "Player": ["Ravi", "Anu", "Kiran", "Meena"],
    "Runs": [40, 60, 80, 100]
}

df = pd.DataFrame(data)

print(df.describe())
```

## SHAPE

### DEFINITION:
The shape attribute in Python's Pandas library is used to find the number of rows and columns in a dataset. It returns a tuple in the format (rows, columns).

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Movie": ["Avatar", "Titanic", "Jaws", "Rocky"],
    "Year": [2009, 1997, 1975, 1976],
    "Rating": [8.0, 7.9, 8.1, 8.2]
}

df = pd.DataFrame(data)

print(df.shape)
```

## COLUMNS

### DEFINITION:

The columns attribute in Python's Pandas library is used to display the names of all columns in a DataFrame. It helps us understand what information is available in a dataset.

#### EXAMPLE:


```python
import pandas as pd

data = {
    "Course": ["Python", "Java", "SQL"],
    "Duration": [30, 45, 20],
    "Fee": [2000, 3000, 1500]
}

df = pd.DataFrame(data)

print(df.columns)
```

## DTYPES

### DEFINITION:

The dtypes attribute in Python's Pandas library is used to display the data type of each column in a DataFrame. It helps us identify whether a column contains integers, decimal numbers, text, or Boolean values.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Item": ["Laptop", "Mouse", "Keyboard"],
    "Price": [45000, 500, 1200],
    "InStock": [True, False, True]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

## MEMORY_USAGE()

### DEFINITION:
The memory_usage() method in Python's Pandas library is used to check how much memory each column of a DataFrame occupies. It helps us understand and optimize the memory used by a dataset.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "City": ["Delhi", "Mumbai", "Chennai"],
    "Population": [1000, 2000, 1500]
}

df = pd.DataFrame(data)

print(df.memory_usage())
```

## NUNIQUE()

### DEFINITION:
The nunique() method in Python's Pandas library is used to count the number of unique (distinct) values in each column of a DataFrame. By default, it does not count missing values (NaN).

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Color": ["Red", "Blue", "Red", "Green", "Blue"],
    "Size": ["S", "XL", "S", "L", "M"]
}

df = pd.DataFrame(data)

print(df.nunique())
```

## VALUE_COUNT()

### DEFINITION:

The value_counts() method in Python's Pandas library is used to count how many times each unique value appears in a column. It helps identify the frequency of values in a dataset.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Payment": ["Cash", "Card", "Cash", "UPI", "Card", "Cash"]
}

df = pd.DataFrame(data)

print(df["Payment"].value_counts())
```
