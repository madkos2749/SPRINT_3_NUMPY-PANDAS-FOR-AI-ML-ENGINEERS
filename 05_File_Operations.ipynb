## CSV FILES

### DEFINITION:
A CSV (Comma-Separated Values) file stores data in a table format, where values are separated by commas. Pandas makes it easy to read data from a CSV file and write data into one.

* pd.read_csv() – reads data from a CSV file.

* df.to_csv() – writes data to a CSV file.

#### EXAMPLE:

```python
import pandas as pd
fruits = {
    'Fruits': ['Apple', 'Banana', 'Carrot', 'Grapes', 'Mango']
}
df = pd.DataFrame(fruits)
df.to_csv("fruits.csv", index = False)
print(fruits)
print ("CSV File ceatetd successfully.!")
```

## EXACEL FILES

### DEFINITION:

Excel files are used to store data in rows and columns. In Python, we can read data from an Excel file, process it, and write the results into a new Excel file using the pandas library.

Excel files store data in worksheets with the extensions .xlsx or .xls.

* 1. **Reading an Excel File:**
     Reading means importing existing Excel data into Python for analysis or processing.

* 2. **Writing an Excel file**
     Writing means saving Python data into a new Excel file or updating the data stored in a file.

#### EXAMPLE:
A manager stores projects list in an Excel files.

```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Aishu", "Rahul", "Priya", "John", "Meena"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["AI", "ML", "DS", "AI", "ML"],
    "Marks": [95, 88, 32, 84, 89]
})

print(students)
students.to_excel("students.xlsx", index = False)
# Read data from Excel
df = pd.read_excel("students.xlsx")

# Select students who scored 40 or more
passed = df[df["Marks"] >= 40]

# Write results to a new Excel file
passed.to_excel("passed_students.xlsx", index=False)

print("Passed students saved successfully!")

print("PASSED STUDENTS LIST IS : ")
print (passed)
```

## JSON FILES

### DEFINITION:

JSON (JavaScript Object Notation) is a format used to store and exchange data in a simple, structured way. In Python, the built-in json module helps us read data from JSON files and write data into them.

1. **Reading a JSON file**
Reading means loading existing JSON data into Python using json.load().

2. **Writing a JSON file**
Writing means saving Python data into a JSON file using json.dump().

#### EXAMPLE:
Scenario: Store a student's details in a JSON file and read them back.

```python
import json

student = {"name": "Anu", "marks": 88}

# Write data to a JSON file
with open("student.json", "w") as file:
    json.dump(student, file)

# Read data from the JSON file
with open("student.json", "r") as file:
    data = json.load(file)

print(data["name"], data["marks"])
```

## PARQUET FILES

### DEFINITION:

Parquet is a column-oriented file format used to store structured data efficiently. It compresses data and is commonly used in big data processing and data analysis.

1. **Reading a Parquet file :**
Reading means loading data from an existing Parquet file into Python using pd.read_parquet().

2. **Writing a Parquet file :**
Writing means saving Python data into a Parquet file using df.to_parquet().

#### EXAMPLE:

Store product prices in a Parquet file and read them back.

```python
import pandas as pd

# Create product data
df = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag"],
    "Price": [10, 50, 300]
})

# Write data to a Parquet file
df.to_parquet("products.parquet", index=False)

# Read data from the Parquet file
data = pd.read_parquet("products.parquet")

print(data)
```

## PICKLE FILES

### DEFINITION:

Pickle is a Python module used to save Python objects into a file and load them back later. It preserves Python objects such as lists, dictionaries, and DataFrames in their original form.

1. **Reading a Pickle file :**
Reading means loading a saved Python object from a file using pickle.load().

2. **Writing a Pickle file :**
Writing means saving a Python object into a file using pickle.dump().

#### EXAMPLE:
Scenario: Save a list of favorite fruits and read it back.

```python
import pickle

fruits = ["Mango", "Apple", "Orange"]

# Write data to a Pickle file
with open("fruits.pkl", "wb") as file:
    pickle.dump(fruits, file)

# Read data from the Pickle file
with open("fruits.pkl", "rb") as file:
    data = pickle.load(file)

print("FAVORITE FRUITS ARE : " , data)
```

## SQL TABLES

### DEFINITION:

SQL tables store data in rows and columns inside a database. Python can connect to a database, read records from a table, and insert new records into it using SQL queries.

1. **Reading an SQL table :**
Reading means retrieving data from a database table using the SQL SELECT statement.

2. **Writing to an SQL table :**
Writing means inserting new data into a table using the SQL INSERT statement.

#### EXAMPLE:

Store and retrieve book details using Python and SQLite.

```python
import sqlite3

# Connect to the database
con = sqlite3.connect("library.db")
cur = con.cursor()

# Create a table
cur.execute("""
CREATE TABLE IF NOT EXISTS books (
    title TEXT,
    price INTEGER
)
""")

# Write data into the table
cur.execute("INSERT INTO books VALUES (?, ?)", ("Python Basics", 250))
con.commit()

# Read data from the table
cur.execute("SELECT * FROM books")
print(cur.fetchall())

con.close()
```

## READING MULTIPLE FILES

### DEFINITION:

Reading multiple files means loading data from several files into a Python program at once. It is useful when data is stored in separate files but needs to be combined and analyzed together.

Python's glob module can find multiple files matching a pattern, and pandas can read and combine their data.

### EXAMPLE:
 Read multiple CSV files containing daily sales and combine them into one dataset.

```python
pd.DataFrame ({"Date" : ["2024-01-01"], "Product":["Groceries"], "Sales" :[100]}).to_csv("sales1.csv", index = False)
pd.DataFrame ({"Date" : ["2024-01-02"], "Product":["Stationary"], "Sales" :[200]}).to_csv("sales2.csv", index = False)
pd.DataFrame ({"Date" : ["2024-01-03"], "Product":["Milk Products"], "Sales" :[350]}).to_csv("sales3.csv", index = False)
print(" 3 Files were Created. ")
```

```python
import pandas as pd
import glob

# Find all CSV files in the folder
files = glob.glob("sales*.csv")

# Read and combine all files
data = pd.concat([pd.read_csv(f) for f in files])

# Save combined data
data.to_csv("all_sales.csv", index=False)

print(data)
```

## FILE PATH HANDLING

### DEFINITION:

File path handling means specifying the location of a file so that Python can read data from it or write data to it correctly.

Python provides the pathlib module to work with file paths easily across different operating systems.

1. **Reading a file path :**
Specifies the location of an existing file from which data needs to be read.

2. **Writing a file path :**
Specifies the location where a new file should be created or data should be saved.

#### EXAMPLE:

Scenario: Read a student's name from a text file and save a greeting in another file.

```python
from pathlib import Path
print(Path("output/greeting.txt").read_text())
# Define file paths
input_file = Path("data/student.txt")
output_file = Path("output/greeting.txt")
input_file.parent.mkdir(parents = True, exist_ok = True)
if not input_file.exists():
    input_file.write_text("Madhuri")
    print("Created data/student.txt with name Madhuri")
# Read data
name = input_file.read_text()

# Create output folder if needed
output_file.parent.mkdir(parents=True, exist_ok=True)

# Write data
output_file.write_text(f"Welcome, {name}!")

print("Greeting saved successfully!")
```
