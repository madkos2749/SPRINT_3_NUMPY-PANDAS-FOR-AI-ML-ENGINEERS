# SELECTING & FILTERING DATA

Selecting and Filtering Data is the process of choosing specific columns or rows from a dataset based on requirements or conditions. In Python, the Pandas library provides methods such as column selection and Boolean filtering to perform these operations.

* **Selecting :** Retrieves specific columns or rows from a dataset.

* **Filtering :** Retrieves rows that satisfy a given condition.

#### EXAMPLE:

Scenario: Select employee names and filter employees whose salary is greater than ₹30,000.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Asha", "Ravi", "Meena", "Arun"],
    "Salary": [25000, 40000, 35000, 28000]
})

# Select one column
print(df["Employee"])

# Filter employees by salary
print(df[df["Salary"] > 30000])
```

## LOC

### DEFINITION:

The loc[] property in Python's Pandas library is used to select specific rows and columns based on their labels. It can also filter rows using conditions.

#### EXAMPLE:

Scenario: Select the details of students who scored more than 70 marks.

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Meena", "Arun"],
    "Marks": [65, 85, 92, 55]
})

# Filter students using loc[]
result = df.loc[df["Marks"] > 70, ["Student", "Marks"]]

print(result)
```

## iLOC

### DEFINITION:
The iloc[] property in Python's Pandas library is used to select specific rows and columns based on their integer positions. Position counting starts from 0.

#### EXAMPLE:

Scenario: Select the first two students and display their names and marks.

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Meena", "Arun"],
    "Marks": [75, 88, 92, 65]
})

# Select first two rows and both columns
result = df.iloc[0:2, 0:2]

print(result)
```

## BOOLEAN INDEXING

### DEFINITION:

Boolean Indexing in Python's Pandas library is a technique used to select rows from a dataset based on conditions that return True or False. Only rows where the condition is True are selected.

#### EXAMPLE:
Scenario: Find products priced below ₹500.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Pen", "Bag", "Book", "Pencil"],
    "Price": [20, 800, 300, 10]
})

# Filter products below ₹500
result = df[df["Price"] < 500]

print(result)
```

## FILTERING ROWS

### DEFINITION:
Filtering Rows in Python's Pandas library means selecting only the rows that satisfy a specific condition. It helps extract relevant records from a dataset for analysis.

#### EXAMPLE:
Scenario: Find students who scored 80 or more marks.

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Anu", "Raj", "Kiran", "Maya"],
    "Marks": [75, 90, 65, 85]
})

# Filter rows based on marks
result = df[df["Marks"] >= 80]

print(result)
```

## QUERY()

### DEFINITION:
The query() method in Python's Pandas library is used to filter rows from a DataFrame by writing conditions as a string. It makes filtering data simple and readable.

#### EXAMPLE:
Scenario: Find movies with a rating greater than 8.

```python
import pandas as pd

df = pd.DataFrame({
    "Movie": ["Leo", "Jawan", "Vikram", "Master"],
    "Rating": [7.5, 7.8, 8.5, 7.2]
})

# Filter movies by rating
result = df.query("Rating > 8")

print(result)
```

## *isin()*

### DEFINITION:
The isin() method in Python's Pandas library is used to check whether values in a column belong to a specified list of values. It returns True for matching values and False for non-matching values.

#### EXAMPLE:
Scenario: Select students who study Python or Java.

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Anu", "Raj", "Maya", "Kiran"],
    "Course": ["Python", "SQL", "Java", "C++"]
})

# Filter students by selected courses
result = df[df["Course"].isin(["Python", "Java"])]

print(result)
```

## BETWEEN()

### DEFINITION:
The between() method in Python's Pandas library is used to filter values that fall within a specified range. By default, it includes both the lower and upper limits.

#### EXAMPLE:
Scenario: Find products priced between ₹100 and ₹500.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Pen", "Bag", "Book", "Watch"],
    "Price": [20, 450, 250, 800]
})

# Filter prices within the range
result = df[df["Price"].between(100, 500)]

print(result)
```

## SORTING DATA

### DEFINITION:

Sorting Data in Python's Pandas library means arranging rows in a specific order based on one or more columns. Data can be sorted in ascending or descending order using the sort_values() method.

* **Ascending order :** Arranges values from smallest to largest.

* **Descending order :** Arranges values from largest to smallest.

#### EXAMPLE:
Scenario: Sort products from the lowest price to the highest price.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [45000, 500, 1500, 12000]
})

# Sort products by price
result = df.sort_values(by="Price")

print(result)
```
