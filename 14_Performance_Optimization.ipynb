# **PERFORMANCE OPTIMIZATION**

Performance optimization means improving the speed and efficiency of Pandas code when working with data. It helps process large DataFrames faster and use less memory.

Common techniques include:

 * Using vectorized operations instead of loops.
 * Selecting only the required columns.
 * Using appropriate data types.
 * Avoiding unnecessary calculations.
 * Using built-in Pandas methods for calculations.

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Chandana", "Sree", "Dilip", "Kedari", "Mallesh"],
    "Salary": [25000, 30000, 40000, 35000, 45000]
})

# Optimized vectorized calculation
employees["Bonus"] = employees["Salary"] * 0.10

print(employees)
```

## VECTORIZATION

### DEFINITION:
* Vectorization is a technique that performs an operation on an entire column or array at once instead of processing each value individually using a loop.

* In Pandas, vectorization makes code shorter, easier to read, and often faster, especially when working with large datasets.

```python
# Vectorization: calculate bonus for all employees at once
employees["Bonus"] = employees["Salary"] * 0.10

print(employees)
```

```python
# Conditional Vectorization
employees["Bonus"] = employees["Salary"] * 0.05
employees["Bonus"] = employees["Bonus"].where(
    employees["Salary"] < 35000,
    employees["Salary"] * 0.10
)

print(employees)
```

## apply()

### DEFINITION:
apply() is a Pandas method used to apply a function to each value in a Series or along an axis of a DataFrame.

#### EXAMPLE:
Using the same employees DataFrame, let's calculate a 10% bonus using both methods.

```python
# Calculate bonus using apply()
employees["Bonus"] = employees["Salary"].apply(
    lambda salary: salary * 0.10
)

print(employees)
```

## Loops()

### DEFINITION:
Loops (for loops) process values one at a time, executing the same instructions repeatedly.

```python
# Calculate bonus using a loop
employees["Bonus"] = 0.0

for i in range(len(employees)):
    employees.loc[i, "Bonus"] = employees.loc[i, "Salary"] * 0.10

print(employees)
```

## MEMORY OPTIMIZATION

### DEFINITION:
* Memory optimization is the process of reducing the amount of RAM used by a Pandas DataFrame without losing important information. It helps handle large datasets more efficiently and can improve processing performance.

Common techniques include:

 * Converting columns to smaller, suitable data types using astype().
 * Using the category data type for repeated text values.
 * Loading only the required columns.
 * Checking memory usage with memory_usage().

```python
# Check memory usage before optimization
print("Before optimization:")
print(employees.memory_usage(deep=True))

# Optimize data types
employees["Name"] = employees["Name"].astype("category")
employees["Salary"] = employees["Salary"].astype("int32")

# Check memory usage after optimization
print("\nAfter optimization:")
print(employees.memory_usage(deep=True))

print("\nData types:")
print(employees.dtypes)
```

## CATEGORY DATA TYPE

### DEFINITION:
* The Category data type in Pandas is used to store columns that contain a limited number of unique values, such as names, cities, departments, or employee roles.

* Instead of storing every repeated text value separately, Pandas stores the unique categories and represents each value using an internal code. This can reduce memory usage when a column contains many repeated values.

```python
employees["Name"] = employees["Name"].astype("category")
print(employees)
```

```python
memory = employees.memory_usage(
    deep=True
).sum()

print(memory)
```

## EFFICIENT PANDAS PRACTICES

### DEFINITION:
Efficient Pandas practices are techniques used to write code that runs faster, uses less memory, and is easier to read and maintain when working with DataFrames.

Common efficient Pandas practices
* **1.Vectorization:** Perform calculations on entire columns instead of using loops.
* **2. Avoid unnecessary loops:** Prefer built-in Pandas operations for calculations.
* **3. Select required columns:** Process only the columns needed for your task.
* **4. Optimize data types:** Use suitable types such as int32 or category when appropriate.
* **5. Handle missing values:** Use methods such as fillna() and dropna() appropriately.
* **6. Avoid unnecessary copies:** Reuse DataFrames when safe instead of creating redundant copies.
* **7. Use efficient filtering:** Apply Boolean indexing to select rows that meet conditions.

```python
selected = employees[
    [
        "Name",
        "Bonus"
    ]
]

print(selected)
```
